```python
import sys
sys.path.insert(0, "/content/ai-hub-models")

from qai_hub_models.models.facemap_3dmm.utils import CachedWebModelAsset, load_numpy

MODEL_ID = "facemap_3dmm"
MODEL_ASSET_VERSION = 1

mean_face = load_numpy(
    CachedWebModelAsset.from_asset_store(MODEL_ID, MODEL_ASSET_VERSION, "meanFace.npy")
)
shape_basis = load_numpy(
    CachedWebModelAsset.from_asset_store(MODEL_ID, MODEL_ASSET_VERSION, "shapeBasis.npy")
)
blendshape_basis = load_numpy(
    CachedWebModelAsset.from_asset_store(MODEL_ID, MODEL_ASSET_VERSION, "blendShape.npy")
)

print(mean_face.shape, shape_basis.shape, blendshape_basis.shape)
import numpy as np

np.save("/content/meanFace.npy", mean_face)
np.save("/content/shapeBasis.npy", shape_basis)
np.save("/content/blendShape.npy", blendshape_basis)

```
